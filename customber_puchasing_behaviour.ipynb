```python
import pandas as pd
file_path = "/Users/chandler/Documents/KARIER/DA/customer_shopping_behavior.csv"
df = pd.read_csv(file_path)

df.head()
```

```python
df.info() #this helps to get the information abut the csv file
```

```python
df.describe(include='all') # df.describe() brings the numerical data while this will bring all  statistics here
```

```python
df.isnull().sum() #this will show null values present in any coloumn like heere it is review rating is zero
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median())) #then for filling these null values we use the
#median of  data to calculate
```

```python
df.isnull().sum()
```

```python
#now we want the column names to be consistent and in a standard format so we will make them smallcase and with underscore if necessary
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python


frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Annually':365,
    'Bi-Weekly':14,
    'Every 3 Months':90,
}

df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```
